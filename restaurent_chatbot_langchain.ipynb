```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    temperature=0
)
```

```python
question = input("Ask something: ")
response = llm.invoke(question)
answer = response.content

if isinstance(answer, list):
    answer = '\n'.join(
        item["text"]
        for item in answer
        if isinstance(item, dict) and item.get("type") == "text"
    )

print(answer)
```

```python
import gradio as gr
from pypdf import PdfReader
from langchain_google_genai import ChatGoogleGenerativeAI

reader = PdfReader("resturaunt.pdf")

restaurent_info = ""

for page in reader.pages:
    text = page.extract_text()
    if text:
        restaurent_info += text + "\n"

print("PDF characters:", len(restaurent_info))

llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    temperature=0
)

def chatbot(message, history):
    print("User:", message)

    prompt = f"""
You are a customer service assistant.
Use the given restaurant information to answer the customer's question.
Only use the provided information.

Restaurant information:
{restaurent_info}

Customer question:
{message}
"""

    print("Sending request to Gemini...")

    response = llm.invoke(prompt)

    print("Gemini responded!")

    return response.content

gradio_app = gr.ChatInterface(fn=chatbot)
```
